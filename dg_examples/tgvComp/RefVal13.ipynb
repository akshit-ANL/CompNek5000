# Reference values - Compressible TGV

```python
import cantera as ct
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print(f"Running Cantera Version: {ct.__version__}")
```

### Set composition

```python
gas = ct.Solution("air1.yaml")
nspec = len(gas.Y)
```

```python
P0 = 1.0e5
T0 = 300
```

```python
Ru = 8.31446261815324
```

```python
Y = np.zeros(nspec)
Y[gas.species_index('O2')] = 0.23
Y[gas.species_index('N2')] = 0.77
gas.TPY = T0, P0, Y
```

```python
print(gas.Y)
```

```python
L0 = 0.00074081/(12.5)
U0 = 34.7878*(12.5)
```

```python
W0 = np.round(gas.mean_molecular_weight/1000,6)
mu0 = np.round(gas.viscosity,9)
rho0 = np.round(gas.density,5)
cp0 = np.round(gas.cp_mass,3)
k0 = np.round(gas.thermal_conductivity,5)
D0 = gas.mix_diff_coeffs[gas.species_index('N2')]
```

```python
# for i in range(len(gas.Y)):
#     print(" i=" + str(i) + " species: " + gas.species_names[i] + " D[i] = " + str(gas.mix_diff_coeffs[i]))
```

```python
Re0 = rho0*L0*U0/mu0
Pr0 = cp0*mu0/k0
Sc0 = mu0/(rho0*D0)
```

```python
print("Reference values ... ")
print("P0    = " + str(P0))
print("T0    = " + str(T0))
print("L0    = " + str(L0))
print("W0    = " + str(W0))
print("mu0   = " + str(mu0))
print("rho0  = " + str(rho0))
print("U0    = " + str(U0))
print("cp0   = " + str(cp0))
print("k0    = " + str(k0))
print("D0    = " + str(D0))
```

```python
print("Non-dimensional parameters ... ")
print("Re0   = " + str(Re0))
print("Pr0   = " + str(Pr0))
print("Sc0   = " + str(Sc0))
```

```python
gamma0 = gas.cp/gas.cv
C0 = np.sqrt(gamma0*Ru*T0*1000/gas.mean_molecular_weight)
Ma0 = U0/C0
print("gamma0 = " + str(gamma0))
print("C0     = " + str(C0))
print("Ma0    = " + str(Ma0))
```

```python
t0 = L0/U0
dt_ndim = 1e-2
dt = dt_ndim*t0
tfinal_ndim = 20
tfinal = tfinal_ndim*t0
print("dt    = " + str(dt))
print("tfinal= " + str(tfinal))
```
